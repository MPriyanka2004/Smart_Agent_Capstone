```python
spark.sql("SHOW TABLES IN default").show()
```

```python
tables = spark.sql("SHOW TABLES IN default").collect()
table_names = [row.tableName for row in tables]

assert "cpg_sales_silver" in table_names, "cpg_sales_silver table missing"

print("Silver table exists")
```

```python
df = spark.table("cpg_sales_silver")
assert df.count() > 0, "silver_sales table is empty"
print("Silver table has data")
```

```python
required_columns = [
    "date", "store_id", "sku_id",
    "units_sold", "revenue", "price",
    "promo_flag"
]

for col in required_columns:
    assert col in df.columns, f"Missing column: {col}"

print("Required columns present")
```
